# 01. Explore ABS Regional Data

This notebook explores the structure of public Australian Bureau of Statistics (ABS) regional datasets used to construct regional income indicators.

Main goals:
1. Load public ABS ASGS and LGA regional datasets
2. Inspect available geography levels
3. Identify income, population, household, and inequality variables
4. Confirm which variables will be used in the income indicator pipeline

```python
import pandas as pd
```

## 1) Download public ABS regional datasets

```python
ASGS_URL = "https://api.data.abs.gov.au/files/ABS_ABS_REGIONAL_ASGS2021_1.0.0.csv"
LGA_URL = "https://api.data.abs.gov.au/files/ABS_ABS_REGIONAL_LGA2021_1.0.0.csv"
```

```python
df_asgs = pd.read_csv(ASGS_URL, low_memory=False)
df_lga = pd.read_csv(LGA_URL, low_memory=False)
```

## 2) Inspect basic structure

```python
print("ASGS dataset shape:", df_asgs.shape)
print("LGA dataset shape:", df_lga.shape)
```

```python
# LGA Column names
df_lga.columns.tolist()
```

### 2.1) Available geography levels in ASGS and LGA data

```python
# ASGS geography data
asgs_geographies = (
    df_asgs[["REGIONTYPE", "Region Type"]]
    .drop_duplicates()
    .sort_values("REGIONTYPE")
)

asgs_geographies
```

```python
# LGA geography data
lga_geographies = (
    df_lga[["REGIONTYPE", "Region Type"]]
    .drop_duplicates()
    .sort_values("REGIONTYPE")
)

lga_geographies
```

```python
# LGA years
df_lga["TIME_PERIOD"].drop_duplicates().sort_values().tolist()
```

## 3) Explore available variables

```python
asgs_vars = (
    df_asgs[["MEASURE", "Data Item"]]
    .drop_duplicates()
    .sort_values("MEASURE")
)

asgs_vars.head(20)
```

### 3.1) Search for income-related variables

```python
income_related_vars = asgs_vars[
    asgs_vars["Data Item"].str.contains(
        "income|household|gini|population|employee|investment|equivalised",
        case=False,
        na=False
    )
].sort_values(["MEASURE", "Data Item"])

income_related_vars
```

### 3.2) Variables selected for the income indicator pipeline

```python
selected_measures = [
    # Population and households
    "ERP_P_2",      # Persons aged 0-4 years
    "ERP_P_3",      # Persons aged 5-9 years
    "ERP_P_4",      # Persons aged 10-14 years
    "ERP_P_20",     # Estimated resident population
    "HHTYPE_5",     # Total households
    "HHTYPE_6",     # Average household size

    # Income components
    "INCOME_3",     # Total employee income
    "INCOME_6",     # Total own unincorporated business income
    "INCOME_9",     # Total investment income

    # Total income
    "INCOME_19",    # Total income earners
    "INCOME_18",    # Total income
    "INCOME_17",    # Median total income
    "INCOME_36",    # Mean total income

    # Equivalised household income
    "EQUIV_2",      # Median equivalised total household income

    # Inequality
    "INCOME_41",    # Gini coefficient
    "INCOME_37",    # P80/P20 ratio
]
```

```python
selected_variable_dictionary = asgs_vars[
    asgs_vars["MEASURE"].isin(selected_measures)
].sort_values("MEASURE")

selected_variable_dictionary
```

## 4.1) Check whether all selected variables are available in ASGS data

```python
available_measures = set(asgs_vars["MEASURE"])
missing_measures = [m for m in selected_measures if m not in available_measures]

print("Missing selected measures:", missing_measures)
```

### 4.1.1) Inspect sample rows for selected variables across key ASGS geographies

```python
sample_asgs = df_asgs[
    (df_asgs["REGIONTYPE"].isin(["AUS", "STE", "SA4", "SA3", "SA2"])) &
    (df_asgs["MEASURE"].isin(selected_measures))
].copy()

sample_asgs.head(20)
```

### 4.1.2) Check observations by geography level

```python
obs_by_geography = (
    sample_asgs
    .groupby("REGIONTYPE")
    .size()
    .reset_index(name="n_observations")
    .sort_values("REGIONTYPE")
)

obs_by_geography
```

### 4.1.3) Inspect SA2 coding structure

```python
sa2_sample = (
    df_asgs[df_asgs["REGIONTYPE"] == "SA2"]
    [["ASGS_2021", "Region"]]
    .drop_duplicates()
    .copy()
)

sa2_sample["ASGS_2021"] = sa2_sample["ASGS_2021"].astype(str).str.zfill(9)

sa2_sample["state_code"] = sa2_sample["ASGS_2021"].str[0]
sa2_sample["sa4_code"] = sa2_sample["ASGS_2021"].str[:3]
sa2_sample["sa3_code"] = sa2_sample["ASGS_2021"].str[:5]
sa2_sample["sa2_code"] = sa2_sample["ASGS_2021"]

sa2_sample.head()
```

## 4.2) Inspect LGA selected variables

```python
lga_vars = (
    df_lga[["MEASURE", "Data Item"]]
    .drop_duplicates()
    .sort_values("MEASURE")
)

lga_selected_variable_dictionary = lga_vars[
    lga_vars["MEASURE"].isin(selected_measures)
].sort_values("MEASURE")

lga_selected_variable_dictionary
```

### 4.2.1) Check whether all selected variables are available in LGA data

```python
available_lga_measures = set(lga_vars["MEASURE"])
missing_lga_measures = [m for m in selected_measures if m not in available_lga_measures]

print("Missing selected LGA measures:", missing_lga_measures)
```

### 4.2.2) Inspect LGA sample rows

```python
sample_lga = df_lga[
    (df_lga["REGIONTYPE"] == "LGA2021") &
    (df_lga["MEASURE"].isin(selected_measures))
].copy()

sample_lga.head(20)
```

## 5) Summary table of selected variables



```python
summary_selected_vars = pd.DataFrame({
    "measure": selected_measures
}).merge(
    asgs_vars.rename(columns={"MEASURE": "measure", "Data Item": "data_item"}),
    on="measure",
    how="left"
)

summary_selected_vars
```
